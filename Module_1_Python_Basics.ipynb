```python
#Function define to calculate accuracy percentage
def calculate_accuracy(correct_trials, total_trials):
  accuracy = (correct_trials / total_trials) * 100
  return accuracy

#Call the function with participant data (e.g., 45 correct out of 50 trials)
participant_accuracy = calculate_accuracy(45, 50)

#Print the result
print("Participant Accuracy (%):", participant_accuracy)
```
